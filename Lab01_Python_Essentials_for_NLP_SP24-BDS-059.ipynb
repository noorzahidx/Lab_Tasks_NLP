## Task 07: String Validation Methods

**(b)**

```python
tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]
for t in tokens:
    if t.isalpha():
        print(t, "-> alphabetic")
    elif t.isdigit():
        print(t, "-> numeric")
    else:
        print(t, "-> mixed")
```

**(c)**

```python
words = ["running", "jumped", "NLP", "ai2025", "data", "42"]
pure_words = []
pure_numbers = []
mixed_tokens = []

for w in words:
    if w.isalpha():
        pure_words.append(w)
    elif w.isdigit():
        pure_numbers.append(w)
    else:
        mixed_tokens.append(w)

print(pure_words)
print(pure_numbers)
print(mixed_tokens)
```

**(d)**

```python
def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"

for t in ["chat", "99", "gpt4", "bert", "2024", "nlp!"]:
    print(t, "->", validate_token(t))
```

## Task 08: Functions and Lambda for NLP Pipelines

**(a)**

```python
def clean(text):
    return text.strip().lower()

clean_lambda = lambda text: text.strip().lower()

sample = "  HELLO Natural Language Processing  "
print(clean(sample))
print(clean_lambda(sample))
```

**(b)**

```python
def tokenize(text):
    return text.split()

def remove_short(tokens, n):
    return [t for t in tokens if len(t) > n]

result = remove_short(tokenize("I love working with NLP tools in Python daily"), 3)
print(result)
```

**(c)**

```python
def word_lengths(tokens):
    return {w: len(w) for w in tokens}

tokens = ["language", "model", "ai", "processing", "deep"]
print(word_lengths(tokens))
```

**(d)**

```python
def nlp_pipeline(text, min_len):
    cleaned = clean(text)
    tokens = tokenize(cleaned)
    tokens = remove_short(tokens, min_len)
    return word_lengths(tokens)

print(nlp_pipeline("Natural Language Processing is a fascinating field", 3))
print(nlp_pipeline("Python makes text processing easy and fun", 2))
```

## Task 09: map(), filter(), sorted() and Counter

**(a)**

```python
words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]
lower_words = list(map(lambda w: w.lower(), words))
word_lens = list(map(lambda w: len(w), words))
print(lower_words)
print(word_lens)
```

**(b)**

```python
words_long = list(filter(lambda w: len(w) > 3, words))
words_with_t = list(filter(lambda w: "t" in w.lower(), words))
print(words_long)
print(words_with_t)
```

**(c)**

```python
alphabetical = sorted(words)
by_length = sorted(words, key=len)
by_length_desc = sorted(words, key=len, reverse=True)
print("Alphabetical:", alphabetical)
print("By length ascending:", by_length)
print("By length descending:", by_length_desc)
```

**(d)**

```python
from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"
freq = Counter(sentence.split())
print(freq)
print(freq.most_common(3))
print(freq["python"])
print(sum(freq.values()), len(freq))
```
