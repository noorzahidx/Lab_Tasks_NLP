## Task 1: Text Cleaning

In this activity, we create a function that takes a paragraph and returns a cleaned list of words.

The words are:
- converted to lowercase
- stripped of leading and trailing punctuation
- returned as a list

```python
paragraph ='''  Hey My name is Noor Zahid.
I am a Data Science Student in 6th semester. 
I am doing my Bachelors from Comsats University Islamabad  '''
```

```python
def cleanData(data):
    lowercased = data.lower()
    strippedSPaces = lowercased.strip()
    words = strippedSPaces.split()

    return words
```

```python
cleanData(paragraph)
```

# Task 2 — Top 10 Most Frequent Words
In this activity, we use Counter to count how many times each word appears in an article.

We then display the 10 most frequent words and their counts.

```python
from collections import Counter

article = """
Python is a popular programming language.
Python is used for data science and machine learning.
Data science uses Python for analysis and visualization.
"""

words = article.lower().split()

word_counts = Counter(words)

top_10 = word_counts.most_common(10)

print("Word\t\tCount")
print("----\t\t-----")

for word, count in top_10:
    print(f"{word}\t\t{count}")
```

## Activity 3: Sentence-Level DataFrame

In this activity, we split a paragraph into individual sentences and create a DataFrame.

Each row represents one sentence.

We calculate:
- word_count: number of words in the sentence
- char_count: number of characters in the sentence

The .apply() function is used to calculate these values.

```python
import pandas as pd

paragraph = """
Python is easy to learn. 
It is widely used in data science. 
Students can use Python for many different projects.
"""

sentences = paragraph.strip().split(".")

sentences = [sentence.strip() for sentence in sentences if sentence.strip()]

df = pd.DataFrame({"sentence": sentences})

df["word_count"] = df["sentence"].apply(lambda x: len(x.split()))
df["char_count"] = df["sentence"].apply(lambda x: len(x))

df
```

## Task 4: Cosine Similarity

Cosine similarity measures how similar two vectors are.

The formula is:

Cosine Similarity = (A · B) / (||A|| × ||B||)

where:
- A · B is the dot product
- ||A|| is the magnitude of vector A
- ||B|| is the magnitude of vector B

A value closer to 1 means the vectors are more similar.
A value closer to 0 means the vectors are less similar.

NumPy is used to perform the calculations.

```python
import numpy as np

A1 = np.array([1, 2, 3])
B1 = np.array([1, 2, 2])

A2 = np.array([1, 0, 3])
B2 = np.array([0, 2, 1])
```

```python
def cosine_similarity(A, B):
    dot_product = np.dot(A, B)
    
    magnitude_A = np.linalg.norm(A)
    magnitude_B = np.linalg.norm(B)
    
    similarity = dot_product / (magnitude_A * magnitude_B)
    
    return similarity
```

```python
similarity_1 = cosine_similarity(A1, B1)
similarity_2 = cosine_similarity(A2, B2)

print("Pair 1:", similarity_1)
print("Pair 2:", similarity_2)
```

### Conclusion

Pair 1 is more similar because its cosine similarity value is higher than Pair 2.

A higher cosine similarity means that the two word-count vectors have a more similar direction.

## Task 5: Review Analysis Using JSON and Counter

In this activity, we load a JSON file containing 10 short reviews.

We:
1. Load the JSON file.
2. Convert the reviews into a DataFrame.
3. Calculate the word count of each review.
4. Use Counter to find the most frequent word in each review.
5. Add the most frequent word as a new column.
6. Sort the DataFrame by word_count in descending order.

```python
import json
import pandas as pd
from collections import Counter
import string

with open("reviews.json", "r") as file:
    data = json.load(file)

df = pd.DataFrame(data)

df.head()
```

```python
df["word_count"] = df["text"].apply(lambda x: len(x.split()))

df.head()
```

```python
def top_word(review):
    words = review.lower().split()
    
    cleaned_words = []
    
    for word in words:
        word = word.strip(string.punctuation)
        cleaned_words.append(word)
    
    counts = Counter(cleaned_words)
    
    return counts.most_common(1)[0][0]

df["top_word"] = df["text"].apply(top_word)

df.head()
```

### Sorting the Reviews

Finally, I sorted the DataFrame by word_count in descending order so that the reviews with the highest number of words appear first.

```python
# sort
df = df.sort_values("word_count", ascending=False)

df.head(10)
```
