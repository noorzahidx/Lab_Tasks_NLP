# Lab Activities

### Activity 1

```python
import spacy

nlp = spacy.load("en_core_web_sm")

# Note: ensure news_paragraph.txt is uploaded or created in the environment
with open("news_paragraph.txt", "r", encoding="utf-8") as f:
    text = f.read()

doc = nlp(text)

print(f"{'Text':<15} | {'POS':<10} | {'TAG'}")
print("-" * 35)
for token in doc:
    print(f"{token.text:<15} | {token.pos_:<10} | {token.tag_}")
```

### Activity 2

```python
target_word = "launch"
for token in doc:
    if token.text.lower() == target_word:
        start = max(0, token.i - 2)
        end = min(len(doc), token.i + 3)
        print(f"Phrase: '... {doc[start:end].text} ...'")
        print(f"Token: {token.text} | POS: {token.pos_}\n")
```

## Activity 3

```python
noise_tags = {"SPACE", "PUNCT", "X"}
print(f"{'Clean Text':<15} | POS")
print("-" * 25)
for token in doc:
    if token.pos_ not in noise_tags:
        print(f"{token.text:<15} | {token.pos_}")
```

## Activity 4

```python
from collections import Counter

for ent in doc.ents:
    print(f"{ent.text:<20} | {ent.label_}")

entity_counts = Counter(ent.label_ for ent in doc.ents)
print("\nEntity Type Counts:")
print("-" * 20)
for ent_type, count in entity_counts.items():
    print(f"{ent_type}: {count}")
```

## Activity 5

```python
print("Before fix:")
for ent in doc.ents:
    if ent.text in ("iPhone", "Pixel 10"):
        print(f"{ent.text:<15} | {ent.label_}")

nlp_custom = spacy.load("en_core_web_sm")
ruler = nlp_custom.add_pipe("entity_ruler", before="ner")
ruler.add_patterns([
    {"label": "PRODUCT", "pattern": "Pixel 10"},
    {"label": "PRODUCT", "pattern": "iPhone"},
])

doc_updated = nlp_custom(text)

print("\nAfter fix (PRODUCT entities):")
for ent in doc_updated.ents:
    if ent.label_ == "PRODUCT":
        print(f"Text: {ent.text:<15} | Label: {ent.label_}")
```

# Home Tasks

## Activity 1

```python
import spacy
import pandas as pd
from collections import Counter, defaultdict

nlp = spacy.load("en_core_web_sm")
df = pd.read_csv("postings.csv")
df = pd.read_csv("postings.csv")

# change these two names if your columns are different
TEXT_COL = "description"
df = df.dropna(subset=[TEXT_COL]).head(100)   # first 100 postings keeps it fast

df["text"] = df[TEXT_COL]
df["id"] = range(1, len(df) + 1)              # simple posting id

docs = list(nlp.pipe(df["text"]))

# ---------- Activity 1: POS tags, top 5 nouns and verbs ----------
print("ACTIVITY 1")
pos_counts = Counter(t.pos_ for d in docs for t in d)
print(pd.DataFrame(pos_counts.most_common(), columns=["POS", "Count"]).to_string(index=False))

nouns = Counter(t.lemma_.lower() for d in docs for t in d if t.pos_ == "NOUN")
verbs = Counter(t.lemma_.lower() for d in docs for t in d if t.pos_ == "VERB")
print("\nTop 5 nouns:", nouns.most_common(5))
print("Top 5 verbs:", verbs.most_common(5))
```

## Activity 2

```python

# ---------- Activity 2: NER grouped by type ----------
print("\nACTIVITY 2")
ent_counts = Counter(ent.label_ for d in docs for ent in d.ents)
for label, count in ent_counts.most_common():
    print(f"{label}: {count}")
print("Most frequent type:", ent_counts.most_common(1)[0][0])
# Most frequent is usually ORG/GPE: every posting names a company and a city.
```

## Activity 3

```python
# ---------- Activity 3: custom EntityRuler for a skill ----------
print("\nACTIVITY 3")
print("Before (Python as entity):")
for i, d in enumerate(docs, 1):
    for ent in d.ents:
        if ent.text == "Python":
            print(f"Posting {i}: {ent.text} | {ent.label_}")

nlp_custom = spacy.load("en_core_web_sm")
ruler = nlp_custom.add_pipe("entity_ruler", before="ner")
ruler.add_patterns([{"label": "SKILL", "pattern": "Python"}])

print("\nAfter (SKILL):")
for i, d in enumerate(nlp_custom.pipe(df["text"]), 1):
    for ent in d.ents:
        if ent.label_ == "SKILL":
            print(f"Posting {i}: {ent.text} | {ent.label_}")
```

## Activity 4

```python
print("\nACTIVITY 4")
target = "design"
for i, d in enumerate(docs, 1):
    for t in d:
        if t.text.lower() == target:
            print(f"Posting {i}: '{d[max(0, t.i-2):t.i+3].text}' | {t.text} | {t.pos_}")
```

## Activity 5

```python

print("\nACTIVITY 5")
# add more skill patterns to the same ruler
ruler.add_patterns([{"label": "SKILL", "pattern": s}
                    for s in ["AWS", "Figma", "SQL", "Docker", "Tableau", "Azure"]])

rows = []
for pid, d in zip(df["id"], nlp_custom.pipe(df["text"])):
    orgs = [e.text for e in d.ents if e.label_ == "ORG"]
    gpes = [e.text for e in d.ents if e.label_ == "GPE"]
    skills = sorted({e.text for e in d.ents if e.label_ == "SKILL"})
    rows.append({
        "posting_id": pid,
        "company": orgs[0] if orgs else "",
        "location": gpes[0] if gpes else "",
        "skills": ", ".join(skills),
    })

result = pd.DataFrame(rows)
print(result.to_string(index=False))
result.to_csv("structured_postings.csv", index=False)
```
